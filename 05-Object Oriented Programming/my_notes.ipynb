## Class Structure

```python
class NameOfClass():
    def __init__(self, param1, param2):
        self.param1 = param1
        self.param2 = param2
    
    def some_method(self):
        # perform some action
        print(self.param1)
```

* self connects the object to the class.
    * self refers to the current object (instance) of the class

```python
class Dog():
    # CLASS OBJECT ATTRIBUTE
    # Define attributes at class-object level
    # The same for any instance of a class
    species = 'mammal'
    
    # Always start with self, it refers to the instance of the class
    def __init__(self, breed, name):
        # Attributes
        # We take in the argument
        # Assign it using self.attribute_name
        self.breed = breed
        self.name = name

    # OPERATIONS/Actions ---> Methods
    # A method is a function that belongs to an object and works with the attributes of that object
    def bark(self, number):
        # We need to say self.name to access the name attribute of the instance
        print(f'WOOF! My name is {self.name} and the number is {number}.')
```

```python
my_dog = Dog('lab', 'Sammy')
```

```python
type(my_dog)
```

```python
my_dog.breed
```

```python
my_dog.species
```

```python
my_dog.bark(5)
```

```python
my_dog.bark
```

```python
class Circle():
    # CLASS OBJECT ATTRIBUTE
    pi = 3.14
    
    def __init__(self, radius=1):
        self.radius = radius
        self.area = radius * radius * Circle.pi
    
    # METHOD
    def get_circumference(self):
        return self.radius * self.pi * 2
```

```python
my_circle = Circle(30)
```

```python
my_circle.pi
```

```python
my_circle.get_circumference()
```

```python
my_circle.area
```

```python
Circle.pi
# You can reassign the class object attribute
```

## Inheritence and Polymorphism

### Inheritence
* A way to form new clases using classes that have already been defined

```python
# BASE CLASS
class Animal():
    def __init__(self):
        print("ANIMAL CREATED")
    
    def who_am_i(self):
        print("I am an animal")
    
    def eat(self):
        print("I am eating")
```

```python
myanimal = Animal()
```

```python
myanimal.who_am_i()
```

```python
class Dog(Animal): # Inherits from Animal
    def __init__(self):
        Animal.__init__(self) # Call the parent class constructor
        print("Dog created")
    
    # Override the who_am_i method
    def who_am_i(self):
        print("I am a dog")
    
    def eat(self):
        print("I am a dog and eating")
    
    def bark(self):
        print("WOOF!")
    
```

```python
mydog = Dog()
```

```python
mydog.who_am_i()
```

```python
mydog.eat()
```

### Polymorphism
* Methods belong to the objects they act on.
* Polymotphism refers to the way in which different object classes can share the same method name. Then, the method can be called from the same place, even though a variety of objects may be passed in.

```python
class Dog():
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        return self.name + " says woof!"
```

```python
class Cat():
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        return self.name + " says meow!"
```

```python
niko = Dog("Niko")
felix = Cat("Felix")
```

```python
print(niko.speak())
```

```python
print(felix.speak())
```

```python
for pet in [niko, felix]:
    print(type(pet))
    # print(type(pet.speak()))
    print(pet.speak())
```

```python
def pet_speak(pet):
    print(pet.speak())
```

```python
pet_speak(niko)
```

```python
pet_speak(felix)
```

### Abstract Classes and Inheritence
* An abstract class is a class that is never expected to be instantiated. It is only meant to be a base class.

```python
class Animal():
    def __init__(self, name):
        self.name = name
    
    # Abstract method: this method is meant to be overridden in subclasses
    def speak(self):
        raise NotImplementedError("Subclass must implement abstract method")
```

```python
myanimal = Animal("fred")
```

```python
myanimal.speak()
```

```python
class Dog(Animal):
    def speak(self):
        return self.name + " says woof!"
```

```python
class Cat(Animal):
    def speak(self):
        return self.name + " says meow!"
```

```python
fido = Dog("Fido")
isis = Cat("Isis")
```

```python
fido.speak()
```

```python
isis.speak()
```

```python
class Book():
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
    
    def __str__(self):
        return f"{self.title} by {self.author}"
    
    def __len__(self):
        return self.pages
    
    def __del__(self):
        print("A book object has been deleted.")
```

```python
# If I don't create a special method, it will return the default representation. In this case, a string representation of the object.
```

```python
b = Book("Python Rocks", "Joe Smith", 200)
```

```python
print(b)
```

```python
str(b)
```

```python
len(b)
```

```python
# Delete an object
del b
```

```python
len(b)
```
