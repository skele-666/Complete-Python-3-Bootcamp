# Errors and Exception Handling

```python
def add (n1, n2):
    print(n1 + n2)
```

```python
add(10, 20)
```

```python
number1 = 10
```

```python
number2 = input("Please enter a number: ")
```

```python
add(number1, number2)
print("Something happened!")
```

```python
try:
    # WANT TO ATTEMPT THIS CODE
    # MAY HAVE AN ERROR
    result = 10 + 10
except:
    # WHAT TO RUN IF THERE IS AN ERROR
    print("Hey it looks like you aren't adding correctly!")
else:
    # WHAT TO RUN IF THERE IS NO ERROR
    # THIS CODE WILL RUN IF THE TRY BLOCK WORKS
    print("Add went well!")
    print(result)
finally:
    # ALWAYS RUN THIS CODE
    print("This will always run at the end!")
```

## Example with File

```python
try:
    # 'w' means write, or create a new file
    f = open('testfile', 'r')
    f.write("Write a test line")
except TypeError:
    print("There was a type error!")
except OSError:
    print("There was an OS error!")
except:
    print("All other exceptions")
finally:
    print("I always run")
    f.close()
```

## Example with Function

```python
def ask_for_int():
    while True:
        try:
            result = int(input("Please provide a number: "))
        except:
            print("Looks like you did not enter an int!")
            continue
        else:
            print("Thank you for providing an int!")
            break
        finally:
            print("I'm going to ask you again!\n")
```

```python
ask_for_int()
```

* Even when break is hit inside the else block, the finally block will still run
* This is useful for closing files or releasing resources
