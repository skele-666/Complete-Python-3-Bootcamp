````python
import random
suits = ('Hearts', 'Diamonds', 'Spades', 'Clubs')
ranks = ('Two', 'Three', 'Four', 'Five', 'Six', 'Seven', 'Eight', 'Nine', 'Ten', 'Jack', 'Queen', 'King', 'Ace')
values = {'Two': 2, 'Three': 3, 'Four': 4, 'Five': 5, 'Six': 6, 'Seven': 7, 'Eight': 8, 'Nine': 9, 'Ten': 10, 'Jack': 11, 'Queen': 12, 'King': 13, 'Ace': 14}
````

## Card Class

````python
class Card:
    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank
        # Why not self.rank?
        self.value = values[rank]
    
    def __str__(self):
        return f"{self.rank} of {self.suit}"
````

## Deck Class
* It will hold a list of Card objects.
* This means the Deck class will return Card class object instances. It's not inheritence or polymorphism.

````python
class Deck:
    def __init__(self):
        self.all_cards = []
        
        # Create a Card object for each suit and rank combination
        for suit in suits:
            for rank in ranks:
                # Create card object
                created_card = Card(suit, rank)
                self.all_cards.append(created_card)
    
    def shuffle(self):
        # shuffle method does not return anything, it just shuffles the deck in place
        random.shuffle(self.all_cards)
    
    def deal_one(self):
        # This method will remove one card from the deck and return it
        return self.all_cards.pop()
````

## Player Class
* Will be used to hold a player's current list of cards.
* A player shoyld be able to add or remove cards from their "hand" (list of Card objects).
* We want them to be able to add a single card or multiple cards to their list.
* When a player plays a card from the "top", it's really `cards.pop(0)`.
* When a player adds a card to the "bottom", it's `cards.append("W")`, as .append() adds to the "bottom" of the list.
* To add multiple cards to the "bottom" of a player's deck, we use cards.extend(new), where new is an array of cards: `new = ["X", "Z"]`

````python
class Player:
    def __init__(self, name):
        self.name = name
        self.all_cards = []  # This will hold the player's cards
    
    # We want players to be able to add or remove cards from their hand
    def remove_one(self):
        return self.all_cards.pop(0) # To remove a card from the "top" of the player's hand   
    
    def add_cards(self, new_cards):
        if type(new_cards) == type([]):
            # Then we are adding multiple cards
            # A list of multiple card objects
            self.all_cards.extend(new_cards)
        else:
            # We are adding a single card
            # For a single card object
            self.all_cards.append(new_cards)

    def __str__(self):
        return f"Player {self.name} has {len(self.all_cards)} cards."
````

## Game Logic
1. Create two instances of Player class
2. Create instance of Deck class (new_deck), shuffle deck
3. Split deck between the two players
4. Check to see if someone has already lost (check for 0 cards), `game_on = True`
    ```
    while game_on
        ...
    ```
5. Each player draws a card, cards are compared.
    * If not a tie, then player wins cards.
    * If a tie, then war. Players must draw additional cards.
        You could be in war for multiple rounds so have war in a while loop
        ```
        while at_war
            ...
        ```
        Whoever wins gets the leftover cards. Check game_on condition.

````python
# GAME SETUP
player_one = Player("One")
player_two = Player("Two")

new_deck = Deck()
new_deck.shuffle()

# Split deck between players
for x in range(26):
    player_one.add_cards(new_deck.deal_one())
    player_two.add_cards(new_deck.deal_one())
    
round_num = 0
game_on = True
````

````python
# While game_on
while game_on:
    round_num += 1
    print(f"Round {round_num}")

    # Check for game over
    if len(player_one.all_cards) == 0:
        print("Player One out of cards! Player Two wins!")
        game_on = False
        break
    elif len(player_two.all_cards) == 0:
        print("Player Two out of cards! Player One wins!")
        game_on = False
        break
    
    # Start new round
    player_one_cards = []
    player_one_cards.append(player_one.remove_one())
    
    player_two_cards = []
    player_two_cards.append(player_two.remove_one())
    
    at_war = True
    # While at_War
    while at_war:
        if player_one_cards[-1].value > player_two_cards[-1].value:
            # Player One wins the round
            player_one.add_cards(player_one_cards)
            player_one.add_cards(player_two_cards)
            at_war = False
        elif player_one_cards[-1].value < player_two_cards[-1].value:
            # Player Two wins the round
            player_two.add_cards(player_two_cards)
            player_two.add_cards(player_one_cards)
            at_war = False
        else:
            print('WAR!')
            
            if len(player_one.all_cards) < 5:
                print("Player One unable to declare war!")
                print("Player Two wins!")
                game_on = False # End the game
                break # End war loop
            elif len(player_two.all_cards) < 5:
                print("Player Two unable to declare war!")
                print("Player One wins!")
                game_on = False
                break
            else:
                for num in range(5):
                    player_one_cards.append(player_one.remove_one())
                    player_two_cards.append(player_two.remove_one())
````
