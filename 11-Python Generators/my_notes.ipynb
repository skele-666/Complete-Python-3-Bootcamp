# Generators
* When a generator function is compiled, they become an object that supports an iteration protocol.
* That means when they are called in your code, they don't actually return a value and then exit.
* They will automatically suspend and resume their execution and state around the last point of value generation.
    * So the generator computes one value and waits until the next value is called for.
* The range() function is a generator

## Creating our own!

```python
def create_cubes(n):
    result = []
    for x in range(n):
        result.append(x**3)
    return result
```

```python
create_cubes(10)
# Keeping this entire list in memory
```

```python
# Unlike here, where we only need the prev. value and the formula to get to the next value
for x in create_cubes(10):
    print(x)
```

```python
def create_cubes(n):
    for x in range(n):
        yield(x**3)
```

```python
for x in create_cubes(10):
    print(x)
# More memory efficient!
```

```python
create_cubes(10)
```

```python
list(create_cubes(10))
```

```python
def gen_fibon(n):
    a = 1
    b = 1
    for i in range(n):
        yield a
        a, b = b, a + b
```

```python
for number in gen_fibon(10):
    print(number)
```

```python
# if it was a normal function...
def gen_fibon(n):
    a = 1
    b = 1
    output = []
    for i in range(n):
        output.append(a)
        a, b = b, a + b
    return output
```

```python
gen_fibon(10)
```

```python
for number in gen_fibon(10):
    print(number)
```

```python
def simple_gen():
    for x in range(3):
        yield x
```

```python
for number in simple_gen():
    print(number)
```
