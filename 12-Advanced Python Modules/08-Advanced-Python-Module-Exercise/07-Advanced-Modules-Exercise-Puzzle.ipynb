```python
zip_obj.extractall('./')
```

```python
# Iterate through each folder and find the phone number
# Formatted ###-###-####
import os
import re
```

```python
phone_pattern = r'\d{3}-\d{3}-\d{4}'

for folder, sub_folders, files in os.walk('./extracted_content'):
    print(f"Currently looking at {folder}")
    print('\n')
    print('The subfolders are: ')
    for sub_fold in sub_folders:
        print(f"\tSubfolder: {sub_fold}")
    
    print('\n')
    print('The files are: ')
    for f in files:
        print(f"\tFile: {f}")
        file_path = os.path.join(folder, f)
        with open(file_path, 'r') as current_file:
            contents = current_file.read()
            match = re.search(phone_pattern, contents)
            if match:
                print(f"Found phone number: {match.group()} in file: {file_path}")
                break
    print('\n')
```
