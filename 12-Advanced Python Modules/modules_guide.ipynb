## shutil

**What you used (detected)**

- `shutil.` attributes: move
- `from shutil import ...`: —

**Why / When to use**

- High‑level file operations: copy/move/remove trees, archives.
- Danger zone: operations overwrite unless you guard (`dirs_exist_ok`).


**Gotchas & habits**

- `rmtree` permanently deletes; prefer `send2trash` in exploratory scripts.
- Copying metadata needs `copy2` not `copy`.


**Mini patterns**

- Zip a folder: `shutil.make_archive('name','zip',root_dir='folder')`
- Mirror a tree (overwrite ok): `copytree(src, dst, dirs_exist_ok=True)`


```py
import shutil
from pathlib import Path

shutil.copy("a.txt", "b.txt")
shutil.copytree("src_dir", "dst_dir", dirs_exist_ok=True)
shutil.move("b.txt", "folder/b.txt")
shutil.make_archive("backup", "zip", root_dir="dst_dir")
```
