### Additional Regex Syntax

```python
re.search(r'cat|dog', 'The cat is here')
# | is or
```

```python
re.findall(r'...at', 'The cat in the hat went splat.')
# . is a wildcard for any character
```

```python
re.findall(r'^\d', '1 is a number, 2 is also a number')
# ^ means "starts with"
```

```python
re.findall(r'\d$', '1 is a number, 2 is also a number. So is 3')
# $ means "ends with"
```

```python
# To exclude characters...
phrase = 'there are 3 numbers 34 inside 5 this sentence'
```

```python
pattern = r'[^\d]+'
# Exclude any digits. + adds all the letters back together
```

```python
re.findall(pattern, phrase)
```

```python
test_phrase = "This is a string! But it has punctuation. How can we remove it?"
```

```python
clean = re.findall(r'[^!.? ]+', test_phrase)
```

## Timing Your Code
### Tracking Time Elapsed

```python
def func_one(n):
    return [str(num) for num in range(n)]
```

```python
func_one(10)
```

```python
def func_two(n):
    return list(map(str, range(n)))
```

```python
func_two(10)
```

```python
import time
```

```python
# CURRENT TIME BEFORE
start_time = time.time()

# RUN CODE
result = func_one(100)

# CURRENT TIME AFTER RUNNING CODE
end_time = time.time()

# ELAPSED TIME
elapsed_time = end_time - start_time

print(elapsed_time)
```

```python
# CURRENT TIME BEFORE
start_time = time.time()

# RUN CODE
result = func_two(100)

# CURRENT TIME AFTER RUNNING CODE
end_time = time.time()

# ELAPSED TIME
elapsed_time = end_time - start_time

print(elapsed_time)
```

### timeit

```python
import timeit
```

```python
stmt = '''
func_one(100)
'''
```

```python
setup = '''
def func_one(n):
    return [str(num) for num in range(n)]
'''
```

```python
# So, you need a statement and a setup.
# The statement is what you want to test.
# The setup is any code you need to run before the statement.
timeit.timeit(stmt, setup, number=1000000)
```

```python
stmt2 = '''
func_two(100)
'''
```

```python
setup2 = '''
def func_two(n):
    return list(map(str, range(n)))
'''
```

```python
timeit.timeit(stmt2, setup2, number=1000000)
```

### %%timeit

```python
%%timeit
func_one(100)
```

```python
%%timeit
func_two(100)
```
