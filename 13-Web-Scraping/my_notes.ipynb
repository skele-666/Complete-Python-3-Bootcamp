# Web Scraping
A general term for techniques involving automating the gathering of data from a website.
## Rules
* Always try to get permission beforehand.
* Too many attempts, your IP address could get blocked.
* Some sites automatically block scraping software.
## Limitations
* Every website is unique, so every web scraping script is unique.
* A slight change/update to a website could completely break your web scraping script.

### Example Project - Book Scraping
<a href="www.toscrape.com">Web Scraping Sandbox</a>

```python
# GOAL: Get title of every book with a 2 star rating
```

```python
# Loop through each page to get all the books
base_url = "http://books.toscrape.com/catalogue/page-{}.html"
```

```python
base_url.format('20')
```

```python
res = requests.get(base_url.format('1'))
```

```python
soup = bs4.BeautifulSoup(res.text, "lxml")
```

```python
len(soup.select(".product_pod")) # 20 books per page
```

```python
products = soup.select(".product_pod")
```

```python
example = products[0]
```

```python
example
```

```python
# Quick and dirty way to get the title
```

```python
'star-rating Three' in str(example)
```

```python
# More robust way to get the title
# looking for star-rating Two
example.select('.star-rating.Three')
```

```python
example.select('.star-rating.Two')
# So I can loop through all the products and check if they are empty or not
```

```python
[] == example.select('.star-rating.Two')
```

```python
# Grab title
example.select('a')[1]['title']
```

```python
# We can check if something is 2 stars (str call in, example.select(rating))
# example.select('a')[1]['title] to grab the book title
```

```python
two_star_titles = []
```

```python
for n in range(1, 51):
    scrape_url = base_url.format(n)
    res = requests.get(scrape_url)
    
    soup = bs4.BeautifulSoup(res.text, "lxml")
    books = soup.select('.product_pod')
    
    # Loop through all the books
    for book in books:
        # If the length of the list is not 0, then it has a 2 star rating
        if len(book.select('.star-rating.Two')) != 0:
            two_star_titles.append(book.select('a')[1]['title'])
```
