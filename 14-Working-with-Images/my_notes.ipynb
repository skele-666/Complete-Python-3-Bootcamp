## Color Transparency
**RGBA - Red, Green, Blue, Alpha**

```python
red = Image.open("red_color.jpg").convert("RGBA")
```

```python
red
```

```python
blue = Image.open("blue_color.png").convert("RGBA")
```

```python
blue
```

```python
blue.putalpha(128)
```

```python
blue
```

```python
red.putalpha(128)
```

```python
red
```

```python
# Mix them together...
blue.paste(im=red, box=(0, 0), mask=red)
```

```python
blue
```

## Saving Images

```python
blue.save("purple.png")
# save() will overwrite
```
