# Emails in Python
## Sending Emails
* Need to connect to an email server, confirm connection, set a protocol, log on and send the message.

```python
import smtplib
```

```python
# Create SMTP object
smtp_object = smtplib.SMTP('smtp.gmail.com', 587)
```

```python
# Run the EHLO method command to greet the server and establish a connection
smtp_object.ehlo()
```

```python
# Initiate TLS email encryotion
smtp_object.starttls()
```

```python
import getpass
```

```python
email = getpass.getpass('Email: ')
password = getpass.getpass('Password: ')
smtp_object.login(email, password)
```

```python
# Send email
from_address = email
to_address = email
subject = input('Enter subject line: ')
message = input('Enter the body message: ')
msg = f'Subject: {subject}\n{message}'

smtp_object.sendmail(from_address, to_address, msg)
```

```python
# Quit connection
smtp_object.quit()
```

```python
import imaplib
```

```python
M = imaplib.IMAP4_SSL('imap.gmail.com')
```

```python
email = getpass.getpass('Email: ')
password = getpass.getpass('Password: ')
```

```python
M.login(email, password)
```

```python
M.list()
```

```python
# Connect to inbox
M.select('inbox')
```

```python
# Search inbox
(typ, data) = M.search(None, 'SUBJECT "I LOVE GREEN BEANS"')
```

```python
typ
```

```python
data
```

```python
email_id = data[0]
```

```python
(result, email_data) = M.fetch(email_id, '(RFC822)')
```

```python
email_data
```

```python
raw_email = email_data[0][1]
```

```python
raw_email_string = raw_email.decode('utf-8')
```

```python
import email
```

```python
email_message = email.message_from_string(raw_email_string)
```

```python
email
```

```python
for part in email_message.walk():
    if part.get_content_type() == 'text/plain':
        body = part.get_payload(decode=True)
        print(body)
```
