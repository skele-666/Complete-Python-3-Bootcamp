# GUIs with Python!
## Interact function/decorator

```python
from ipywidgets import interact, interactive, fixed
import ipywidgets as widgets
```

```python
# Using interact to create a simple slider widget
def func(x):
    return x
```

```python
interact(func, x=10)
```

```python
interact(func, x=True)
```

```python
interact(func, x='Hello')
```

```python
@interact(x=True, y=fixed(1.0))
def g(x, y):
    return (x, y)
```

## Widget Abbreviations

```python
interact(func, x=widgets.IntSlider(min=-100, max=100, step=1, value=0));
```

```python
interact(func, x=(-10.0, 10.0, 0.01))
# This is an abbreviation for the IntSlider widget above
```

```python
@interact(x=(0.0, 20.0, 0.5))
def h(x=5.0):
    return x
```

```python
# Dropdown widget
interact(func, x=['hello', 'option 2', 'option 3'])
```

```python
interact(func, x={'one': 10, 'two': 20})
```

### Widget Styling and Layout

```python
import ipywidgets as widgets
from IPython.display import display
```

```python
w = widgets.IntSlider()
display(w)
```

```python
w.layout.margin = 'auto'
w.layout.height = '75px'
```

```python
x = widgets.IntSlider(value=15, description='New Slider')
display(x)
```

```python
x.layout = w.layout
```

```python
widgets.Button(description='Ordinary Button', button_style='info')
```

```python
b1 = widgets.Button(description='Custom Colour')
b1.style.button_color = 'lightgreen'
b1
```

```python
b1.style.keys
```

```python
b2 = widgets.Button(description='NEW')
b2.style = b1.style
b2
```

```python
s1 = widgets.IntSlider(description='My Handle')
s1.style.handle_color = 'lightblue'
s1
```
